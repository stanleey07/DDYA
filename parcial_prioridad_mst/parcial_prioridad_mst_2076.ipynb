## 5) Análisis de complejidad

### Justificación
El max heap permite organizar las tareas csegun su prioridad. Kruskal permite conectar los seis dispositivosutilizando las conexiones de menos distacia y evitando ciclos, y segun la complejidad O(n), mientras que en la inserccion y extraccion tienen una complejidad maxima de O(log n), debido a los intercambios que se realizan por niveles de arbol, y estos dos algoritmos permiten resolver las dos necesidad de la empesa de una menra mas eficiente.



### **FUNCION T[n]**
Max-Heap:
T(n) = n + 3 log₂(n)
Complejidad: O(n).
Kruskal:
T(E) = E + E log₂(E)
Complejidad: O(E log E).
Donde n representa la cantidad de tareas y E la cantidad de conexiones posibles. En este problema hay inicialmente 6 tareas y 15 conexiones posibles.

### **Mejor y peor caso**
En el mejor caso, las operaciones individuales del heap no necesitan intercambios y las aristas ya están ordenadas, permitiendo seleccionar las conexiones sin muchos descartes. En el peor caso, las operaciones individuales recorren la altura del heap, las aristas requieren ordenamiento completo y se revisan conexiones que pueden generar ciclos. Las gráficas siguientes muestran los modelos teóricos para ambas etapas.

## 6) Código funcional
La implementación completa está organizada en los cinco puntos del enunciado. Ejecutar las celdas en orden. Se usan únicamente Python, `math` para las distancias y `matplotlib` para las visualizaciones; las operaciones del heap y el algoritmo de Kruskal se implementan manualmente.

```python
import math
import matplotlib.pyplot as plt
from itertools import combinations


def max_heapify(heap, i):
    while True:
        mayor = i
        izquierdo = 2*i + 1
        derecho = 2*i + 2
        if izquierdo < len(heap) and heap[izquierdo][2] > heap[mayor][2]:
            mayor = izquierdo
        if derecho < len(heap) and heap[derecho][2] > heap[mayor][2]:
            mayor = derecho
        if mayor == i:
            break
        heap[i], heap[mayor] = heap[mayor], heap[i]
        i = mayor


def build_max_heap(heap):
    for i in range(len(heap)//2 - 1, -1, -1):
        max_heapify(heap, i)


def extract_max(heap):
    if not heap:
        raise IndexError('Cola vacía')
    tarea = heap[0]
    ultimo = heap.pop()
    if heap:
        heap[0] = ultimo
        max_heapify(heap, 0)
    return tarea


def insertar(heap, tarea):
    heap.append(tarea)
    i = len(heap)-1
    while i > 0:
        padre = (i-1)//2
        if heap[padre][2] >= heap[i][2]:
            break
        heap[padre], heap[i] = heap[i], heap[padre]
        i = padre


def es_max_heap(heap):
    for i in range(1, len(heap)):
        if heap[(i-1)//2][2] < heap[i][2]:
            return False
    return True


def dibujar_heap(heap, titulo):
    plt.figure(figsize=(8, 4))
    posiciones = []
    for i in range(len(heap)):
        nivel = int(math.log2(i+1))
        lugar = i - (2**nivel-1)
        x = (lugar+0.5)/2**nivel
        y = -nivel
        posiciones.append((x,y))
        if i > 0:
            xp, yp = posiciones[(i-1)//2]
            plt.plot([xp,x], [yp,y], 'k-')
    for i in range(len(heap)):
        x,y = posiciones[i]
        plt.scatter(x,y,s=1800,color='lightblue',zorder=3)
        plt.text(x,y,f'{heap[i][0]}\n{heap[i][2]}',ha='center',va='center',zorder=4)
    plt.xlim(0,1)
    plt.ylim(-max(1, int(math.log2(len(heap))))-0.5,0.5)
    plt.title(titulo)
    plt.axis('off')
    plt.show()


tareas = [
    ('T01','Configurar servidor',6),
    ('T02','Reparar sensor',10),
    ('T03','Actualizar software',4),
    ('T04','Revisar cámara',8),
    ('T05','Generar respaldo',3),
    ('T06','Configurar alarma',7)
]
heap = tareas.copy()
print('Arreglo inicial:', heap)
build_max_heap(heap)
print('Max-Heap:', heap)
dibujar_heap(heap,'Max-Heap inicial')

procesadas = []
for i in range(2):
    tarea = extract_max(heap)
    procesadas.append(tarea)
    print('Tarea procesada:', tarea)
    print('Heap:',heap)
    assert es_max_heap(heap)
insertar(heap,('T07','Falla en sistema contra incendios',12))
print('Heap final:',heap)
dibujar_heap(heap,'Max-Heap final')

dispositivos = {'A':(2,2),'B':(7,3),'C':(4,7),'D':(10,8),'E':(13,3),'F':(15,7)}
nombres = {'A':'Servidor','B':'Cámara','C':'Sensor','D':'Alarma','E':'Router','F':'Panel'}
ids = list(dispositivos)
aristas = []
for i in range(len(ids)):
    for j in range(i+1,len(ids)):
        a,b = ids[i],ids[j]
        x1,y1 = dispositivos[a]
        x2,y2 = dispositivos[b]
        peso = math.sqrt((x2-x1)**2+(y2-y1)**2)
        aristas.append((peso,a,b))
        print(a,'-',b,':',round(peso,2))


def dibujar_dispositivos(conexiones,titulo):
    plt.figure(figsize=(9,5))
    for peso,a,b in conexiones:
        x1,y1 = dispositivos[a]
        x2,y2 = dispositivos[b]
        plt.plot([x1,x2],[y1,y2],'g-')
        plt.text((x1+x2)/2,(y1+y2)/2,f'{peso:.2f}',bbox={'facecolor':'white'})
    for a in dispositivos:
        x,y = dispositivos[a]
        plt.scatter(x,y,color='blue')
        plt.annotate(a+' - '+nombres[a],(x,y),xytext=(4,7),textcoords='offset points')
    plt.title(titulo)
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.xlim(0,18)
    plt.ylim(0,10)
    plt.gca().set_aspect('equal')
    plt.grid()
    plt.show()


dibujar_dispositivos([], 'Posición de los dispositivos')


def encontrar(padres,a):
    if padres[a] != a:
        padres[a] = encontrar(padres,padres[a])
    return padres[a]


def kruskal(ids,aristas):
    padres = {a:a for a in ids}
    rangos = {a:0 for a in ids}
    mst = []
    for peso,a,b in sorted(aristas):
        raiz_a = encontrar(padres,a)
        raiz_b = encontrar(padres,b)
        if raiz_a != raiz_b:
            if rangos[raiz_a] < rangos[raiz_b]:
                padres[raiz_a] = raiz_b
            else:
                padres[raiz_b] = raiz_a
                if rangos[raiz_a] == rangos[raiz_b]:
                    rangos[raiz_a] += 1
            mst.append((peso,a,b))
        if len(mst) == len(ids)-1:
            break
    return mst


print('Aristas ordenadas:',sorted(aristas))
mst = kruskal(ids,aristas)
total = sum(peso for peso,a,b in mst)
for peso,a,b in mst:
    print(a,'-',b,':',round(peso,2))
print(f'Longitud total: {total:.2f}')
dibujar_dispositivos(mst,f'MST - Longitud total: {total:.2f}')

n = list(range(2,101))
plt.figure()
plt.plot(n,[k+3 for k in n],label='Mejor caso')
plt.plot(n,[k+3*math.log2(k) for k in n],label='Peor caso')
plt.title('Complejidad Max-Heap (modelo teórico)')
plt.xlabel('Tareas')
plt.ylabel('Costo relativo')
plt.legend()
plt.grid()
plt.show()
e = [k*(k-1)//2 for k in n]
plt.figure()
plt.plot(n,[2*m for m in e],label='Mejor caso (aristas ordenadas)')
plt.plot(n,[2*m+m*math.log2(m) for m in e],label='Peor caso')
plt.title('Complejidad del grafo (modelo teórico)')
plt.xlabel('Dispositivos')
plt.ylabel('Costo relativo')
plt.legend()
plt.grid()
plt.show()


def conectado(ids,aristas):
    visitados = {ids[0]}
    for i in ids:
        for peso,a,b in aristas:
            if a in visitados or b in visitados:
                visitados.update([a,b])
    return visitados == set(ids)


prueba = tareas.copy()
build_max_heap(prueba)
assert es_max_heap(prueba) and prueba[0][0] == 'T02'
print('Test 1 OK')
assert extract_max(prueba)[0] == 'T02' and es_max_heap(prueba)
print('Test 2 OK')
assert extract_max(prueba)[0] == 'T04' and es_max_heap(prueba)
print('Test 3 OK')
assert es_max_heap(heap) and heap[0][0] == 'T07' and len(heap) == 5
print('Test 4 OK')
assert len(aristas) == 15 and len({(a,b) for peso,a,b in aristas}) == 15
print('Test 5 OK')
assert len(mst) == 5 and conectado(ids,mst)
print('Test 6 OK')
minimo = min(sum(p for p,a,b in grupo) for grupo in combinations(aristas,5) if conectado(ids,grupo))
assert math.isclose(total,minimo)
print('Test 7 OK')
vacio = []
try:
    extract_max(vacio)
except IndexError:
    pass
else:
    raise AssertionError('La cola vacía debe generar un error')
insertar(vacio,('X','Prueba',1))
assert extract_max(vacio)[0] == 'X' and vacio == []
print('Test 8 OK')
print('8 de 8 tests aprobados')
```
