### 5) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación
- Función T[n]
- Mejor y peor caso con gráfica



1. **JUSTIFICACION**:
- El algoritmo utiliza recursividad para calcular la cantidad de suscriptores de un determinado mes.
Cuando n es mayor o igual a 2, la función realiza dos llamadas recursivas:
T(n-1) y T(n-2)
Por esta razón, los mismos valores pueden calcularse varias veces. A medida que aumenta n, también aumenta rápidamente la cantidad de llamadas que debe realizar el algoritmo.

- Por ejemplo, para calcular T(5): T(5)=T(4)+T(3)
y para calcular cada uno de esos valores se vuelven a realizar nuevas llamadas. Por lo tanto, el algoritmo presenta un crecimiento exponencial en su tiempo de ejecución.

2. **FUNCION T[n]**:
- T(n) = T(n-1)+T(n-2)+c 
- Con los casos base: T(0)=c y T(1)=c 
- T(n)= T(n-1)+T(n-2). n>=2.

3. **MEJOR CASO Y PEOR CASO (con grafica)**:

- **Mejor caso**: El mejor caso ocurre cuando el usuario ingresa 0 o 1, porque son los casos base de la funcion. El programa retorna inmediatamente el resultado y no realiza llamadas recursivas: T(n) = O(1)

- **Peor caso**: El peor caso ocurre cuando n>= 2 porque la funcion realiza dos llamadas recursivas: T(n) = T(n-1)+T(n-2). Y se repiten calculos. Por esto, el tiempo de ejecucion crece eponencialmente: T(n) = O(2^n).


## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def suscriptores(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return suscriptores(n - 1) + suscriptores(n - 2)


n = int(input("Ingrese el número del mes: "))

if n < 0:
    print("El número del mes debe ser mayor o igual a 0.")
else:
    resultado = suscriptores(n)
    print("En el mes", n, "habrá", resultado, "nuevos suscriptores.")
```
