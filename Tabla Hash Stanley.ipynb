## 1) Descripción del problema **[1 pts]**



Una tienda necesita mejorar la organización y administración de los productos registrados en su inventario, debido a que la búsqueda de información mediante estructuras lineales puede resultar ineficiente cuando aumenta la cantidad de productos almacenados.

Actualmente, localizar un producto requiere recorrer diferentes registros hasta encontrar el código correspondiente, lo que puede incrementar el tiempo de búsqueda y dificultar la gestión del inventario.

Para solucionar esta problemática, se propone desarrollar un sistema basado en una **Tabla Hash**, utilizando el código numérico de cada producto como llave de identificación. La posición de almacenamiento se determinará mediante la función hash:

$$
h(código) = código \bmod m
$$

Donde:

- **código:** identificador único de cada producto.
- **m:** tamaño de la Tabla Hash.
- **h(código):** posición calculada para almacenar o localizar el producto.

Cada producto estará compuesto por cuatro atributos: código, nombre, precio y cantidad disponible.

Además, debido a que diferentes códigos pueden generar la misma posición, se implementará el método de **encadenamiento** para solucionar las colisiones y permitir que varios productos compartan una misma posición de la tabla.

El sistema deberá permitir registrar, buscar, eliminar y actualizar productos, así como visualizar el contenido completo de la estructura.

El propósito principal es implementar una solución que permita administrar el inventario de manera organizada y eficiente, aprovechando las ventajas de las Tablas Hash para reducir el tiempo de acceso a los productos.

### 5) Análisis de complejidad **[15 pts]**
Incluye:

Para analizar la eficiencia del sistema de inventario, se utiliza la notación Big O, teniendo en cuenta el comportamiento de las operaciones principales de la Tabla Hash.

### Variables utilizadas

- **m:** tamaño de la Tabla Hash, es decir, número de posiciones disponibles.
- **n:** cantidad total de productos almacenados.
- **k:** cantidad de productos almacenados en una misma posición debido a las colisiones.

La función hash utilizada es:

$$
h(código) = código \bmod m
$$

El cálculo de esta función tiene una complejidad constante, ya que requiere una única operación de módulo.

### Complejidad de las operaciones

| Operación | Mejor caso | Peor caso |
|---|---:|---:|
| Insertar producto | O(1) | O(n) |
| Buscar producto | O(1) | O(n) |
| Eliminar producto | O(1) | O(n) |
| Actualizar cantidad | O(1) | O(n) |
| Mostrar toda la tabla | O(m + n) | O(m + n) |

### Análisis de cada operación

**1. Inserción:** En el mejor caso, el producto se almacena directamente en su posición y no existen otros elementos que revisar. En el peor caso, todos los productos se encuentran en una misma cadena y es necesario recorrerla para comprobar que el código no esté duplicado.

**2. Búsqueda:** En el mejor caso, el producto se encuentra al inicio de la cadena. En el peor caso, es necesario recorrer todos los productos almacenados en esa posición o comprobar que el producto no existe.

**3. Eliminación:** En el mejor caso, el producto se encuentra al inicio de la cadena y se elimina directamente. En el peor caso, se debe recorrer la cadena completa para localizarlo.

**4. Actualización:** En el mejor caso, el producto se encuentra al inicio de la cadena. En el peor caso, se debe recorrer toda la cadena para encontrarlo y modificar su cantidad.

**5. Visualización:** El sistema debe recorrer las `m` posiciones de la tabla y mostrar los `n` productos almacenados. Por esta razón, su complejidad es O(m + n).

### Complejidad espacial

La complejidad espacial es:

$$
O(m+n)
$$

Esto se debe a que se necesita espacio para las `m` posiciones de la Tabla Hash y para almacenar los `n` productos registrados, incluyendo aquellos que se encuentran en las cadenas de encadenamiento.

### Conclusión

La Tabla Hash permite realizar las operaciones de inserción, búsqueda, eliminación y actualización con una complejidad promedio cercana a **O(1)**, siempre que los productos se distribuyan adecuadamente entre las posiciones disponibles.

Sin embargo, cuando se presentan muchas colisiones y los productos se concentran en una misma cadena, las operaciones pueden alcanzar una complejidad de **O(n)** en el peor caso.

Por lo tanto, el rendimiento del sistema depende de una distribución adecuada de los códigos dentro de la Tabla Hash y del manejo correcto de las colisiones mediante encadenamiento.


## 6) Código funcional **[15 pts]**

```python

class Producto:
    def __init__(self, codigo, nombre, precio, cantidad):
        self.codigo = codigo
        self.nombre = nombre
        self.precio = precio
        self.cantidad = cantidad


class TablaHash:
    def __init__(self, m):
        if not isinstance(m, int) or isinstance(m, bool) or m <= 0:
            raise ValueError("El tamaño de la tabla debe ser un entero positivo.")

        self.m = m
        self.tabla = [[] for _ in range(m)]

    def funcion_hash(self, codigo):
        return codigo % self.m

    def insertar(self, codigo, nombre, precio, cantidad):
        indice = self.funcion_hash(codigo)

        for producto in self.tabla[indice]:
            if producto.codigo == codigo:
                return False

        nuevo_producto = Producto(codigo, nombre, precio, cantidad)
        self.tabla[indice].append(nuevo_producto)
        return True

    def buscar(self, codigo):
        indice = self.funcion_hash(codigo)

        for producto in self.tabla[indice]:
            if producto.codigo == codigo:
                return producto

        return None

    def eliminar(self, codigo):
        indice = self.funcion_hash(codigo)

        for posicion, producto in enumerate(self.tabla[indice]):
            if producto.codigo == codigo:
                self.tabla[indice].pop(posicion)
                return True

        return False

    def actualizar(self, codigo, nueva_cantidad):
        if not isinstance(nueva_cantidad, int) or isinstance(nueva_cantidad, bool):
            return False

        if nueva_cantidad < 0:
            return False

        producto = self.buscar(codigo)

        if producto is not None:
            producto.cantidad = nueva_cantidad
            return True

        return False

    def mostrar_producto(self, producto):
        print("\n" + "-" * 45)
        print("          INFORMACIÓN DEL PRODUCTO")
        print("-" * 45)
        print(f"Código:    {producto.codigo}")
        print(f"Nombre:    {producto.nombre}")
        print(f"Precio:    ${producto.precio:,.2f}")
        print(f"Cantidad:  {producto.cantidad} unidades")
        print("-" * 45)

    def mostrar(self):
        print("\n" + "=" * 65)
        print("                 CONTENIDO DE LA TABLA HASH")
        print("=" * 65)

        for indice in range(self.m):
            print(f"\nPosición {indice} -> ", end="")

            if len(self.tabla[indice]) == 0:
                print("Vacía")
            else:
                for posicion, producto in enumerate(self.tabla[indice]):
                    print(
                        f"[Código: {producto.codigo} | "
                        f"Nombre: {producto.nombre} | "
                        f"Precio: ${producto.precio:,.2f} | "
                        f"Cantidad: {producto.cantidad}]",
                        end=""
                    )

                    if posicion < len(self.tabla[indice]) - 1:
                        print(" -> ", end="")

                print()

        print("\n" + "=" * 65)

```
