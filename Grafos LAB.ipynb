### 5) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación
- Función T[n]
- Mejor y peor caso con gráfica

Complejidad de BFS
BFS utiliza una cola y visita los vértices y conexiones de la red.
Cada vértice se visita como máximo una vez y cada conexión se revisa durante el recorrido.
Por lo tanto:
T(V,E) = O(V + E)
La complejidad de BFS es:
O(V + E)
5.2 Complejidad de DFS
DFS utiliza una pila y recorre los vértices y conexiones del grafo.
Cada vértice se visita como máximo una vez y las conexiones son revisadas durante el recorrido.
Por lo tanto:
T(V,E) = O(V + E)
La complejidad de DFS es:
O(V + E)
5.3 Complejidad de búsqueda de un centro
La búsqueda de un centro utiliza BFS.
Mejor caso
El centro buscado es el mismo centro desde el cual comienza el recorrido.
Por lo tanto:
T(V,E) = O(1)
Peor caso
Es necesario recorrer todos los vértices y conexiones para encontrar el centro o comprobar que no existe una ruta.
Por lo tanto:
T(V,E) = O(V + E)
5.4 Complejidad al agregar un centro
Si se agrega un nuevo centro con k conexiones, se deben registrar esas conexiones.
Por lo tanto:
T(k) = O(k)
La complejidad depende directamente de la cantidad de conexiones que tenga el nuevo centro.

## 6) Código funcional **[15 pts]**

```python

grafo = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "G"],
    "F": ["C", "G"],
    "G": ["E", "F"]
}


def mostrar_grafo():
    for centro in grafo:
        print(centro, "->", grafo[centro])


def bfs(inicio):
    if inicio not in grafo:
        return []

    cola = [inicio]
    visitados = {inicio}
    recorrido = []
    niveles = {inicio: 0}
    posicion = 0

    while posicion < len(cola):
        actual = cola[posicion]
        posicion += 1

        recorrido.append(actual)

        for vecino in grafo[actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                niveles[vecino] = niveles[actual] + 1
                cola.append(vecino)

    return recorrido, niveles


def dfs(inicio):
    if inicio not in grafo:
        return []

    pila = [inicio]
    visitados = set()
    recorrido = []

    while pila:
        actual = pila.pop()

        if actual not in visitados:
            visitados.add(actual)
            recorrido.append(actual)

            for vecino in reversed(grafo[actual]):
                if vecino not in visitados:
                    pila.append(vecino)

    return recorrido


def buscar_ruta(inicio, destino):
    if inicio not in grafo or destino not in grafo:
        return None

    cola = [inicio]
    visitados = {inicio}
    padres = {inicio: None}
    posicion = 0

    while posicion < len(cola):
        actual = cola[posicion]
        posicion += 1

        if actual == destino:
            ruta = []
            nodo = destino

            while nodo is not None:
                ruta.append(nodo)
                nodo = padres[nodo]

            return ruta[::-1]

        for vecino in grafo[actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                padres[vecino] = actual
                cola.append(vecino)

    return None


def agregar_centro():
    nuevo = input("Nuevo centro: ").strip().upper()

    if nuevo in grafo:
        print("El centro ya existe.")
        return

    if nuevo == "":
        print("Nombre invalido.")
        return

    conexiones = input(
        "Conexiones separadas por coma: "
    ).upper().split(",")

    conexiones = [c.strip() for c in conexiones if c.strip()]

    if not conexiones:
        print("Debe agregar al menos una conexion.")
        return

    for conexion in conexiones:
        if conexion not in grafo:
            print("El centro", conexion, "no existe.")
            return

        if conexion == nuevo:
            print("No puede conectarse consigo mismo.")
            return

    grafo[nuevo] = []

    for conexion in conexiones:
        if conexion not in grafo[nuevo]:
            grafo[nuevo].append(conexion)

        if nuevo not in grafo[conexion]:
            grafo[conexion].append(nuevo)

    print("Centro agregado correctamente.")


def mostrar_bfs(inicio):
    resultado = bfs(inicio)

    if not resultado:
        print("Centro no encontrado.")
        return

    recorrido, niveles = resultado

    print("BFS:", " -> ".join(recorrido))

    for nivel in sorted(set(niveles.values())):
        centros = [
            centro for centro in recorrido
            if niveles[centro] == nivel
        ]

        print("Nivel", nivel, ":", ", ".join(centros))


def mostrar_dfs(inicio):
    recorrido = dfs(inicio)

    if not recorrido:
        print("Centro no encontrado.")
        return

    print("DFS:", " -> ".join(recorrido))


def buscar_centro():
    inicio = input("Centro inicial: ").strip().upper()
    destino = input("Centro que desea buscar: ").strip().upper()

    ruta = buscar_ruta(inicio, destino)

    if ruta is None:
        print("No existe una ruta.")
    else:
        print("Centro encontrado.")
        print("Ruta:", " -> ".join(ruta))
        print("Cantidad de conexiones:", len(ruta) - 1)


def ejecutar_pruebas():
    assert bfs("A")[0] == [
        "A", "B", "C", "D", "E", "F", "G"
    ]

    assert bfs("A")[1] == {
        "A": 0,
        "B": 1,
        "C": 1,
        "D": 2,
        "E": 2,
        "F": 2,
        "G": 3
    }

    assert dfs("A") == [
        "A", "B", "D", "E", "G", "F", "C"
    ]

    assert buscar_ruta("A", "F") == ["A", "C", "F"]

    assert buscar_ruta("A", "G") == ["A", "B", "E", "G"]

    assert bfs("Z") == ([], {})

    assert dfs("Z") == []

    print("Todas las pruebas pasaron.")


def menu():
    while True:
        print("\n1. Mostrar grafo")
        print("2. BFS")
        print("3. DFS")
        print("4. Buscar centro")
        print("5. Agregar centro")
        print("6. Salir")

        opcion = input("Opcion: ")

        if opcion == "1":
            mostrar_grafo()

        elif opcion == "2":
            inicio = input("Centro inicial: ").upper()
            mostrar_bfs(inicio)

        elif opcion == "3":
            inicio = input("Centro inicial: ").upper()
            mostrar_dfs(inicio)

        elif opcion == "4":
            buscar_centro()

        elif opcion == "5":
            agregar_centro()
            mostrar_grafo()

            inicio = input(
                "Centro para ejecutar BFS y DFS: "
            ).upper()

            mostrar_bfs(inicio)
            mostrar_dfs(inicio)

        elif opcion == "6":
            print("Programa finalizado.")
            break

        else:
            print("Opcion invalida.")


ejecutar_pruebas()
menu()
```
