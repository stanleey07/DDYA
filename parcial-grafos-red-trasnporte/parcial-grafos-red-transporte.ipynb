5. ANÁLISIS DE COMPLEJIDAD

5.1 JUSTIFICACIÓN
------------------------------------------------------------
Para resolver el problema de la red de estaciones se utiliza un
gráfico representado mediante una lista de adyacencia.
Los principales algoritmos utilizados son BFS y DFS.
Para el análisis se define:
n = número de estaciones de la red.
En el problema, cada estación representa un vértice y cada
conexión representa un arista.
La cantidad de operaciones realizadas por los algoritmos depende
principalmente de la cantidad de estaciones y conexiones que
deben ser recorridas.
BFS utiliza una cola para recorrer la red por niveles, mientras
que DFS utiliza una pila o recursividad para recorrer una rama
hasta donde sea posible.
En ambos casos, una estación se visita solamente una vez, tal
como lo establece el enunciado.

5.2 FUNCIÓN T[n]
------------------------------------------------------------
Para expresar la cantidad de operaciones en función del tamaño
de la entrada, se define:
T(n) = cantidad de operaciones realizadas para recorrer una red
      de n estaciones.
Para el recorrido de la red, cada estación debe ser procesada y
sus conexiones deben ser revisadas.
Por lo tanto, la función de tiempo puede expresarse de forma
general como:
T(n) = c1(n) + c2(n) + c3
Agrupando las constantes:
T(n) = an + b
Por lo tanto, la función de crecimiento es:
T(n) = O(n)
Esto significa que el tiempo de ejecución aumenta de manera
lineal a medida que aumenta la cantidad de estaciones que deben
ser procesadas.

5.3 MEJOR CASO
------------------------------------------------------------
El mejor caso se presenta cuando el recorrido puede completarse
realizando la menor cantidad posible de operaciones.
Para la búsqueda de una ruta mediante BFS, el mejor caso ocurre
cuando la estación de destino está directamente conectada con la
estación de origen.
Ejemplo:
Central → Norte
En este caso, BFS encuentra el destino después de procesar una
cantidad mínima de estaciones.
Por lo tanto:
T_mejor(n) = O(1)
Para un recorrido completo de BFS o DFS, sin embargo, se deben
visitar todas las estaciones de la red, por lo que el recorrido
completo mantiene una complejidad:
T_mejor(n) = O(n)

5.4 PEOR CASO
------------------------------------------------------------
El peor caso ocurre cuando el algoritmo debe recorrer
prácticamente toda la red antes de encontrar el resultado o
determinar que no existe una ruta.
Por ejemplo, si se busca una estación que se encuentra al final
del recorrido, BFS debe procesar las estaciones anteriores hasta
alcanzarla.
También puede ocurrir que la estación destino no sea alcanzable.
En ese caso, el algoritmo debe recorrer todas las estaciones
posibles para comprobar que no existe una ruta.
Por lo tanto:
T_peor(n) = O(n)

5.5 COMPARACIÓN
------------------------------------------------------------

| Situación | Complejidad |

| Búsqueda - mejor caso | O(1) |
| Búsqueda - peor caso | O(n) |
| BFS - recorrido completo | O(n) |
| DFS - recorrido completo | O(n) |

## 6) Código funcional **[15 pts]**

```python
from collections import deque

class GrafoEstaciones:

    def agregar_estacion(self, estacion):
        if estacion not in self.grafo:
            self.grafo[estacion] = []
            return True
        return False

    def agregar_conexion(self, estacion1, estacion2):
        if estacion1 not in self.grafo:
            self.agregar_estacion(estacion1)

        if estacion2 not in self.grafo:
            self.agregar_estacion(estacion2)

        if estacion2 not in self.grafo[estacion1]:
            self.grafo[estacion1].append(estacion2)

        if estacion1 not in self.grafo[estacion2]:
            self.grafo[estacion2].append(estacion1)

    def mostrar_grafo(self):
        print("LISTA DE ADYACENCIA")

        for estacion, vecinos in self.grafo.items():
            print(f"{estacion}: {', '.join(vecinos)}")

    def bfs(self, inicio):
        if inicio not in self.grafo:
            return None, None

        cola = deque([inicio])
        visitados = {inicio}
        niveles = {inicio: 0}
        orden = []

        while cola:
            actual = cola.popleft()
            orden.append(actual)

            for vecino in self.grafo[actual]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    niveles[vecino] = niveles[actual] + 1
                    cola.append(vecino)

        return orden, niveles

    def dfs(self, inicio):
        if inicio not in self.grafo:
            return None

        visitados = set()
        orden = []

        def recorrer(actual):
            visitados.add(actual)
            orden.append(actual)

            for vecino in self.grafo[actual]:
                if vecino not in visitados:
                    recorrer(vecino)

        recorrer(inicio)

        return orden

    def buscar_ruta(self, origen, destino):
        if origen not in self.grafo or destino not in self.grafo:
            return None

        cola = deque([origen])
        visitados = {origen}
        padre = {origen: None}

        while cola:
            actual = cola.popleft()

            if actual == destino:
                break

            for vecino in self.grafo[actual]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    padre[vecino] = actual
                    cola.append(vecino)

        if destino not in padre:
            return None

        ruta = []
        actual = destino

        while actual is not None:
            ruta.append(actual)
            actual = padre[actual]

        ruta.reverse()
        return ruta

    def mostrar_bfs(self, inicio):
        orden, niveles = self.bfs(inicio)

        if orden is None:
            print(f"Error: la estación '{inicio}' no existe.")
            return

        print(f"\nBFS DESDE {inicio}")
        print(" -> ".join(orden))

        print("Niveles:")

        for estacion in orden:
            print(f"Nivel {niveles[estacion]}: {estacion}")

    def mostrar_dfs(self, inicio):
        orden = self.dfs(inicio)

        if orden is None:
            print(f"Error: la estación '{inicio}' no existe.")
            return

        print(f"\nDFS DESDE {inicio}")
        print(" -> ".join(orden))

    def mostrar_ruta(self, origen, destino):
        ruta = self.buscar_ruta(origen, destino)

        print(f"\nRUTA {origen} -> {destino}")

        if ruta is None:
            print(f"No existe una ruta entre {origen} y {destino}.")
            return

        print("Ruta encontrada:")
        print(" -> ".join(ruta))
        print(f"Cantidad de conexiones: {len(ruta) - 1}")


red = GrafoEstaciones()
red.grafo = {}

estaciones = [
    "Central",
    "Norte",
    "Museo",
    "Parque",
    "Universidad",
    "Hospital",
    "Terminal"
]

for estacion in estaciones:
    red.agregar_estacion(estacion)

conexiones = [
    ("Central", "Norte"),
    ("Central", "Museo"),
    ("Norte", "Parque"),
    ("Norte", "Universidad"),
    ("Museo", "Hospital"),
    ("Universidad", "Terminal"),
    ("Hospital", "Terminal")
]

for estacion1, estacion2 in conexiones:
    red.agregar_conexion(estacion1, estacion2)

print("RED INICIAL")
red.mostrar_grafo()

red.mostrar_bfs("Central")

red.mostrar_dfs("Central")

print("\nRECORRIDOS DESDE UNIVERSIDAD")
red.mostrar_bfs("Universidad")
red.mostrar_dfs("Universidad")

red.mostrar_ruta("Central", "Terminal")

print("\nPRUEBA DE ESTACIÓN INEXISTENTE")
red.mostrar_bfs("Aeropuerto")

red.agregar_estacion("Aeropuerto")
red.agregar_conexion("Aeropuerto", "Terminal")
red.agregar_conexion("Aeropuerto", "Hospital")

print("\nRED ACTUALIZADA")
red.mostrar_grafo()

red.mostrar_bfs("Aeropuerto")

red.mostrar_dfs("Aeropuerto")

red.mostrar_ruta("Aeropuerto", "Central")

red.agregar_conexion("Aeropuerto", "Terminal")

print("\nPRUEBA DE CONEXIÓN DUPLICADA")
red.mostrar_grafo()

print("\nPROGRAMA FINALIZADO")
```
